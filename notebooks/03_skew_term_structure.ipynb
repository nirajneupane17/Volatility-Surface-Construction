```python
surf = pd.read_csv("../data/vol_surface.csv")
term = pd.read_csv("../data/term_structure.csv")
print("Maturity | ATM IV | Risk Rev | Butterfly")
for Td in [7,30,60,90,180,360]:
    rr = risk_reversal(surf, Td)
    bf = butterfly(surf, Td)
    row = term[term["maturity_days"]==Td]
    if len(row): print(f"  {Td:3d}d  {row.atm_iv_pct.values[0]:.2f}%  {rr:+.2f}vp  {bf:.2f}vp")
```
