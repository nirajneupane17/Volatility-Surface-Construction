```python
import sys; sys.path.insert(0,"../src")
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from svi_model import calibrate_svi, svi_implied_vol, svi_rmse, butterfly_arbitrage_check
print("OK")
```

```python
opts=pd.read_csv("../data/options_chain.csv")
for exp in ["30D","90D","365D"]:
    d=opts[opts["expiry"]==exp].sort_values("log_moneyness")
    T=d["T"].values[0]; k=d["log_moneyness"].values; iv=d["implied_vol"].values
    params=calibrate_svi(k,iv,T,n_restarts=20)
    if params:
        rmse=svi_rmse(k,iv,T,params)*100
        bf=butterfly_arbitrage_check(k,params,T)
        print(f"{exp}: RMSE={rmse:.4f}%  butterfly_arb_free={bf}")
```
